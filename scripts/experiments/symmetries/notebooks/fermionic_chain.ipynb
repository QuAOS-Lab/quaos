```python
from sympleq.core.symmetries import pauli_reduce, min_qudit_clifford_symmetry, qudit_cost, find_clifford_symmetries
from sympleq.models.fermionic_chain import fermionic_chain_hamiltonian
from sympleq.core.circuits.gate_decomposition_to_circuit import gate_to_circuit
import numpy as np
import time
```

```python
n = 16
J = 1
V = 2
D = 1
D_vec = np.random.uniform(-D/2, D/2, size=n)
periodic = False
H = fermionic_chain_hamiltonian(n, J, V, D_vec, periodic=periodic)
h_red, conditioned_hamiltonians, C_F, phases = pauli_reduce(H)

print(conditioned_hamiltonians[0])
# C_F.show()
```

```python
H_next = conditioned_hamiltonians[0]

F, S, T = min_qudit_clifford_symmetry(H_next)
assert H_next.to_standard_form() == F.act(H_next).to_standard_form()
```

```python
print(F.symplectic)
print(qudit_cost(S))
print(H_next.n_qudits())
```

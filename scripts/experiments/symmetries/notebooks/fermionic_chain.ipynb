```python
from sympleq.core.symmetries import pauli_reduce, min_qudit_clifford_symmetry, qudit_cost
from sympleq.models.fermionic_chain import fermionic_chain_hamiltonian
from sympleq.core.circuits.gate_decomposition_to_circuit import gate_to_circuit
import numpy as np
import time
```

```python
n = 3
J = 1
V = 2
D = 1
D_vec = np.random.uniform(-D/2, D/2, size=n)
periodic = False
H = fermionic_chain_hamiltonian(n, J, V, D_vec, periodic=periodic)

print(H)
```

```python
F, S, T = min_qudit_clifford_symmetry(H)
assert H.to_standard_form() == F.act(H).to_standard_form()
```

```python
print(F.symplectic)
print(qudit_cost(F))
```

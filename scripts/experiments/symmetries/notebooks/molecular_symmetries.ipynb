```python
import time
from sympleq.models.chemistry import water_molecule, beh2_molecule, ch4_molecule, c2h4_molecule, c6h6_molecule
from sympleq.core.circuits import SWAP
from sympleq.core.symmetries.clifford import find_clifford_symmetries, qudit_cost, min_qudit_clifford_symmetry
from sympleq.core.circuits import Circuit
import numpy as np
```

```python
# Water molecule
H = c2h4_molecule(basis='sto-3g')
print(H.n_paulis())
print(H.n_qudits())

start = time.time()
F, T, S = min_qudit_clifford_symmetry(H)
end = time.time()
assert H.to_standard_form() == F.act(H).to_standard_form()
H_prime = T.inv().act(H)
assert H_prime == S.act(H_prime)
print(f'Found minimal symmetry in {end - start} seconds')

```

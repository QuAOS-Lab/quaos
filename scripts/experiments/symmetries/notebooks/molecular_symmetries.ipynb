```python
import time
from sympleq.models.chemistry import water_molecule, beh2_molecule, ch4_molecule, c2h4_molecule, c6h6_molecule
from sympleq.core.circuits import SWAP
from sympleq.core.symmetries.clifford import find_clifford_symmetries, qudit_cost, min_qudit_clifford_symmetry
from sympleq.core.circuits import Circuit
import numpy as np
```

```python
# Water molecule
H = water_molecule(basis='sto-3g')
print(H.n_paulis())
print(H.n_qudits())

start = time.time()
F, T, S = min_qudit_clifford_symmetry(H, circuit_augmented_graph='wl')
end = time.time()
assert H.is_close(F.act(H, tuple([2]*H.n_qudits())), literal=False), "Found symmetry does not stabilize H"
# H_prime = T.inverse().act(H)
# assert H_prime.is_close(S.act(H_prime), literal=False)
print(f'Found minimal symmetry in {end - start} seconds')

```
